# Futures / Spot Pair Backtest

This notebook loads the two supplied market-data files and runs a simple two-leg spread backtest through the repository's `Backtest.backtest_trading_day` engine. The default assignment is file 1 short and file 2 long; change the parameters below to reverse it.

The CSV contents alone do not verify whether a file is futures or spot. The notebook reports observable diagnostics and keeps that identity assignment explicit.

```python
from pathlib import Path
import sys
import types

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

REPO_ROOT = Path.cwd().resolve()
if REPO_ROOT.name == 'notebooks':
    REPO_ROOT = REPO_ROOT.parent
if not (REPO_ROOT / 'backtest.py').exists():
    REPO_ROOT = Path(r'c:\Users\tommy\backtest_fut_dual')

# The checkout uses common.* imports while its files live at the repository root.
try:
    import common.backtest as backtest_module
except ModuleNotFoundError:
    common_package = types.ModuleType('common')
    common_package.__path__ = [str(REPO_ROOT)]
    sys.modules['common'] = common_package
    import common.backtest as backtest_module

from common.backtest import Backtest, load
from common.foundation_contracts import InstrumentSpec
from common.market import Market
from common.sessions import DEFAULT_SESSION_CALENDAR
from common.strategy import Strategy

DATA_DIR = REPO_ROOT / 'data'
SHORT_FILE = DATA_DIR / 'market_data_2024-01-25_1.csv'
LONG_FILE = DATA_DIR / 'market_data_2024-01-25_2.csv'

# Explicit role assignment. Set SHORT_FILE/LONG_FILE to reverse the experiment.
SHORT_PRODUCT = 'SHORT_LEG'
LONG_PRODUCT = 'LONG_LEG'
TICK = 1.0
MULTIPLIER = 1.0
FEE_LOT = 0.0
ENTRY_Z = 1.0
EXIT_Z = 0.0
ORDER_QTY = 1
print('Repository:', REPO_ROOT)
print('Short file:', SHORT_FILE.name)
print('Long file: ', LONG_FILE.name)
```

## Inspect the two inputs

```python
raw_short = pd.read_csv(SHORT_FILE, parse_dates=['timestamp', 'exchtime'])
raw_long = pd.read_csv(LONG_FILE, parse_dates=['timestamp', 'exchtime'])

diagnostics = pd.DataFrame({
    'file': [SHORT_FILE.name, LONG_FILE.name],
    'rows': [len(raw_short), len(raw_long)],
    'first_exchtime': [raw_short.exchtime.min(), raw_long.exchtime.min()],
    'last_exchtime': [raw_short.exchtime.max(), raw_long.exchtime.max()],
    'mean_mid': [(raw_short.bidpx0 + raw_short.askpx0).mean() / 2, (raw_long.bidpx0 + raw_long.askpx0).mean() / 2],
    'mean_totalvol_delta': [raw_short.totalvol.diff().fillna(raw_short.totalvol.iloc[0]).mean(), raw_long.totalvol.diff().fillna(raw_long.totalvol.iloc[0]).mean()],
    'mean_openinterest': [raw_short.openinterest.mean(), raw_long.openinterest.mean()],
})
display(diagnostics)
print('Both files have the same quote schema and 70 rows in this sample.')
print('The larger volume/open-interest values in file 2 are clues, not proof that it is futures.')
```

```python
# Use the repository's existing data-cleaning and derived-trade-price logic.
short_frame = load(str(SHORT_FILE), MULTIPLIER, TICK, calendar=DEFAULT_SESSION_CALENDAR)
long_frame = load(str(LONG_FILE), MULTIPLIER, TICK, calendar=DEFAULT_SESSION_CALENDAR)
short_frame = short_frame.sort_index()
long_frame = long_frame.sort_index()

short_mid = (short_frame.bidpx0 + short_frame.askpx0) / 2.0
long_mid = (long_frame.bidpx0 + long_frame.askpx0) / 2.0
aligned = pd.concat({'short_mid': short_mid, 'long_mid': long_mid}, axis=1).sort_index().ffill().dropna()
aligned['spread'] = aligned['long_mid'] - aligned['short_mid']
spread_mean = aligned['spread'].expanding(min_periods=2).mean()
spread_std = aligned['spread'].expanding(min_periods=2).std(ddof=0).replace(0, np.nan)
aligned['spread_z'] = ((aligned['spread'] - spread_mean) / spread_std).fillna(0.0)

# Mean reversion: long the spread when it is cheap, short it when expensive.
aligned['spread_target'] = 0
aligned.loc[aligned['spread_z'] <= -ENTRY_Z, 'spread_target'] = 1
aligned.loc[aligned['spread_z'] >= ENTRY_Z, 'spread_target'] = -1
aligned['spread_target'] = aligned['spread_target'].replace(0, np.nan).ffill().fillna(0).astype(int)
aligned.loc[aligned['spread_z'].abs() <= EXIT_Z, 'spread_target'] = 0

# Leg positions: target +1 on the long leg and -1 on the short leg.
target_by_product = {
    SHORT_PRODUCT: (-aligned['spread_target']).to_dict(),
    LONG_PRODUCT: aligned['spread_target'].to_dict(),
}
display(aligned.head())
```

## Run through `Backtest`

```python
class TargetPositionStrategy(Strategy):
    def __init__(self, name, market, target_positions):
        super().__init__(name, market, MULTIPLIER, TICK, None, 0.0, fee_lot=FEE_LOT, parallel=True, record_diagnostics=True)
        self.target_positions = target_positions
        self.last_target = 0

    def quote(self, md):
        # Signals are keyed by exchange time; decisions are still processed by receive time.
        timestamp = pd.Timestamp(md.get('exchange_ts', md['datetime']))
        target = int(self.target_positions.get(timestamp, self.last_target))
        if self.session_step_count >= self.session_total_steps:
            target = 0
        delta = target - self.pos
        if delta:
            price = md['askpx0'] if delta > 0 else md['bidpx0']
            order = self.submit_order(price, delta, aggressive=True, reason_code='target_position')
            # The repository matcher requires an explicit order id for immediate aggressive execution.
            fills = self.market.match(self.contract, order['order_id']) if order is not None else []
            if fills:
                self.num_fills += len(fills)
                self.filled_qty += sum(abs(fill['qty']) for fill in fills)
                self.update_pos(fills)
        self.last_target = target

market = Market(mult=MULTIPLIER, tick=TICK)
short_strategy = TargetPositionStrategy(SHORT_PRODUCT, market, target_by_product[SHORT_PRODUCT])
long_strategy = TargetPositionStrategy(LONG_PRODUCT, market, target_by_product[LONG_PRODUCT])

specs = {
    SHORT_PRODUCT: InstrumentSpec(SHORT_PRODUCT, TICK, MULTIPLIER, DEFAULT_SESSION_CALENDAR, 'notional-fee-per-lot', 'manual-pair'),
    LONG_PRODUCT: InstrumentSpec(LONG_PRODUCT, TICK, MULTIPLIER, DEFAULT_SESSION_CALENDAR, 'notional-fee-per-lot', 'manual-pair'),
}
combined = pd.concat([
    short_frame.assign(contract=SHORT_PRODUCT),
    long_frame.assign(contract=LONG_PRODUCT),
], axis=0).sort_values('timestamp', kind='stable')

engine = Backtest(market, [short_strategy, long_strategy], MULTIPLIER, TICK, parallel=True, instrument_specs=specs)
summary = engine.backtest_trading_day(
    combined, specs, trading_day=aligned.index[0].date(), strategy_products=[SHORT_PRODUCT, LONG_PRODUCT]
)
summary
```

```python
per_leg = pd.DataFrame({
    SHORT_PRODUCT: short_strategy.summarize(),
    LONG_PRODUCT: long_strategy.summarize(),
}).T
combined_summary = pd.Series({
    'pnl': per_leg['pnl'].sum(),
    'fees': per_leg['fees'].sum(),
    'net_pnl': per_leg['net_pnl'].sum(),
    'num_orders': per_leg['num_orders'].sum(),
    'num_fills': per_leg['num_fills'].sum(),
    'filled_qty': per_leg['filled_qty'].sum(),
})
display(per_leg[['pnl', 'fees', 'net_pnl', 'num_orders', 'num_fills', 'filled_qty']])
display(combined_summary.to_frame('combined'))
```

```python
records = pd.concat([
    pd.DataFrame(short_strategy.session_record),
    pd.DataFrame(long_strategy.session_record),
], ignore_index=True)
records['datetime'] = pd.to_datetime(records['datetime'])
combined_pnl = records.groupby('datetime', sort=True)['pnl'].sum().rename('combined_pnl')
# Diagnostics are recorded at receive time; map them to the nearest exchange-time sample.
combined_pnl.index = combined_pnl.index.round('1ms')
combined_pnl = combined_pnl.groupby(level=0).last()
combined_pnl = combined_pnl.reindex(
    aligned.index, method='nearest', tolerance=pd.Timedelta(milliseconds=100)
).ffill().fillna(0.0)

fig, axes = plt.subplots(2, 1, figsize=(13, 7), sharex=True)
aligned[['short_mid', 'long_mid']].plot(ax=axes[0], title='Short-leg and long-leg mid prices')
axes[0].set_ylabel('price')
combined_pnl.plot(ax=axes[1], color='black', drawstyle='steps-post', marker='o', title='Overall cumulative PnL: short one leg + long the other')
axes[1].axhline(0.0, color='gray', linewidth=0.8)
axes[1].set_ylabel('PnL')
plt.tight_layout()
plt.show()
```

## Interpretation

This is a tiny 70-row demonstration, not evidence that either file is futures or spot and not a meaningful trading result. Before using the result for research, verify instrument metadata, contract multipliers, tick sizes, fee and rebate schedules, session calendars, and use a materially longer sample with out-of-sample evaluation.
